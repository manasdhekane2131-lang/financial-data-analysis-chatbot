```python
import pandas as pd

df = pd.read_csv("financial_data.csv")

df
```

```python
df = df.sort_values(["Company", "Year"])

df["Revenue Growth (%)"] = (
    df.groupby("Company")["Total Revenue"].pct_change() * 100
)

df
```

```python
df["Net Income Growth (%)"] = (
    df.groupby("Company")["Net Income"].pct_change() * 100
)

df
```
